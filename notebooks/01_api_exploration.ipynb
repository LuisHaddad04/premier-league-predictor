# TESTING API #1: APIFOOTBALL

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv("../.env")

API_KEY = os.getenv("API_FOOTBALL_KEY")

print("API key loaded:", API_KEY is not None)
```

```python
# This is the API-Football server and having the /fixtures tells us what information we're looking at, so in this case it's fixtures.
# Endpoints are sort of like different doors into the API
url = "https://v3.football.api-sports.io/fixtures"

# API needs to know we have permission to use it, so we're sending the API key along with the request
# "Here's my API key"
headers = {
    "x-apisports-key": API_KEY
}

# Parameters are set to tell the API exactly what we want.
# API-Football assigns leagues ID Numbers, so the EPL is 39.
# We're looking at historical data from the season "2025" (2025/2026)
params = {
    "league": 39,
    "season": 2024
}

# This sort of acts as Python contacting API-Football.
# "Sending a GET request to the fixtures endpoint of the server and authenticating it using the API key and requesting EPL fixtures for the season 2025."
# Python stores that response in the variable "response".
response = requests.get(
    url,
    headers = headers,
    params = params
)

# Used to check whether the request worked or not.
# Status code: 200 means a success.
print("Status code:", response.status_code)
```

```python
# When an API sends information back, it is sent in JSON format.
# JSON is a standardized way for applications to exchange structured data.
# JSON format looks like: 
# {
    #"league": "Premier League",
    #"season": 2025,
    #"fixtures": [...]
# } 

# Receives the JSON API response and converts into Python objects we can work with.
data = response.json()

# Outer object becomes a Python dictionary.
# <class 'dict'>
print(type(data))

# This shows us the top-level categories the API returned.
print(data.keys())
```

```python
print("Endpoint:", data["get"])
print("Parameters:", data["parameters"])
print("Errors:", data["errors"])
print("Number of results:", data["results"])
```

```python
# "data" is the entire API response which contains 380 matches, so having the "[0]" is saying that we just want to look at the first match on the list.

first_match = data["response"][0]

first_match
```

```python
# Using another API endpoint which is used for specific match-level performance stats.
# /fixtures/statistics.
stats_url = "https://v3.football.api-sports.io/fixtures/statistics"

# Our parameter set for this is the specific fixture ID for the match we want to look at.
stats_params = {
    "fixture": 1208021
}

# Authenticated GET request.
stats_response = requests.get(
    stats_url, 
    headers = headers,
    params = stats_params
)

# Converts the API response into Python data.
stats_data = stats_response.json()

print("Status code:", stats_response.status_code)
print("Errors:", stats_data["errors"])
print("Number of results:", stats_data["results"])
```

```python
# Our previous code was looking at games that haven't occured yet.
# Let's try looking at completed games so far this season.
completed_url = (
    "https://api.thestatsapi.com/api/football/matches"
    "?competition_id=comp_3039"
    "&season_id=sn_8406098"
    "&date_to=2026-08-26"
)

completed_response = requests.get(
    completed_url,
    headers = test_headers
)

print("Status code:", completed_response.status_code)
print("Request URL:", completed_response.url)
```

```python
# We're looking at how many matches have been completed so far.
# As of right now (August 26th, 2026, there should only be 10 matches completed).
completed_data = completed_response.json()

print("Matches returned:", len(completed_data["data"]))
print("Meta:", completed_data["meta"])
```

```python
for match in completed_data["data"]:
    print(
        match["matchday"],
        "-",
        match["utc_date"],
        "-",
        match["home_team"]["name"],
        "vs",
        match["away_team"]["name"],
        "-",
        match["score"]["final_score"],
        "-",
        match["status"],
        "- xG available:",
        match["xg_available"]
    )
```
